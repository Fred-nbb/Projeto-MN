**Import do pacote + parâmetros do problema:**

```python
from metodos_numericos import bisseccao, posicao_falsa, ponto_fixo, newton_raphson, secante
import math

g = 9.81
L = 4.0
t = 2.5
v_alvo = 5.0

def f(H):
    r = math.sqrt(2 * g * H)
    v = r * math.tanh((r / (2 * L)) * t)
    return v - v_alvo

def fdev(H):
    r = math.sqrt(2 * g * H)
    u = r * t / (2 * L)
    sech2 = 1 - math.tanh(u)**2
    return (g / r) * math.tanh(u) + (g * t / (2 * L)) * sech2

def phi(H):
    r = math.sqrt(2 * g * H)
    u = r * t / (2 * L)
    return (v_alvo / math.tanh(u))**2 / (2 * g)
```

**Rodar os 5 métodos:**

```python
eps = 1e-4

resultados = {
    "Bisseção":       bisseccao(f, 1, 2, eps),
    "Posição Falsa":  posicao_falsa(f, 1, 2, eps),
    "Ponto Fixo":     ponto_fixo(phi, f, 1.5, eps, eps),
    "Newton-Raphson": newton_raphson(f, fdev, 1.5, eps, eps),
    "Secante":        secante(f, 1, 2, eps, eps),
}

print(f"{'Método':<16}{'H encontrado':<16}{'Iterações':<10}")
print("-" * 42)
for nome, (raiz, hist) in resultados.items():
    print(f"{nome:<16}{raiz:<16.6f}{len(hist):<10}")
```
